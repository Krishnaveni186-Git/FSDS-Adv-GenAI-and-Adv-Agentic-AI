# Bitwise Operator: Bitwise operator works on binary number i.e 0 & 1. Computer understands this internally numbers in binary form.

# There are 6 types of Bitwise operator:

# 1. Compliment ~
# 2. And &
# 3. OR |
# 4. Xor (xclusive ^)
# 5. Left Shift << 
# 6. Right Shift >> 

```python
# 1. Compliment (~): It is always negative value (-) 

# Ex: ~12 = -13
```
