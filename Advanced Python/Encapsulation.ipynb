# Encapsulation: Encapsulation restricts direct access to class attributes and methods.

Q. Why Use Encapsulation?
- Encapsulation provides several benefits:

- **Data Protection**: Prevents accidental modification of data
- **Validation**: You can validate data before setting it
- **Flexibility**: Internal implementation can change without affecting external code
- **Control**: You have full control over how data is accessed and modified

```python
class car:
    def __init__(self):
        self.mil = 10
        self.brand = 'BMW'

c1 = car()
c2 = car()

print(c1.mil, c1.brand)
print(c2.mil, c2.brand)

print(id(c1))
print(id(c2))
```

# Instance Method : The method which access instance variable. 

```python
class student:
    school = 'nit'

    def __init__(self, m1, m2, m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):  # instance method
        return(self.m1 + self.m2 + self.m3)/3

s1 = student(10, 20, 30)
s2 = student( 20, 50, 80)

print(s1.avg())
print(s2.avg())

print(s1.m1)
print(s2.m2)
```

# Class Method: Method which acces class variable

```python
class student:
    
    school = 'nit'

    def __init__(self, m1, m2, m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   
        return(self.m1 + self.m2 + self.m3)/3

    def info(cls):    # class method
        return cls.school

s1 = student(10, 20, 30)
s2 = student( 20, 50, 80)

print(s1.avg())
print(s2.avg())

print(s1.info())
print(s2.info())
```

```python
class student:
    
    school = 'nit'

    def __init__(self, m1, m2, m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   
        return(self.m1 + self.m2 + self.m3)/3
        
    @classmethod
    def info(cls):    # class method
        return cls.school

s1 = student(10, 20, 30)
s2 = student( 20, 50, 80)

s1.info()
```

# Static Method:  Method which doesnot use any variable i.e. class or instance.

```python
class student:
    
    school = 'nit'

    def __init__(self, m1, m2, m3):
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3

    def avg(self):   
        return(self.m1 + self.m2 + self.m3)/3


    @classmethod
    def getSchool(cls):   
        return cls.school
    
    @staticmethod
    def info():   
        print('This is Student Class')

s1 = student(10, 20, 30)
s2 = student( 20, 50, 80)

print(s1.avg())
print(s2.avg())

print(s1.getSchool())
print(s2.getSchool())

s1.info()
s2.info()
```
