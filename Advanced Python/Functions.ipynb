# 3 - Default Argument: Default argument: Default argument use a predefined value when no value is passed during the function call.

# 4 Variable length argument: A variable-length argument (also called variadic argument) is a function parameter that allows a function to accept any number of arguments.

```python
def sum(a,b):
    c = a + b
    print(c)
sum(5,6)
```

```python
def sum(a,b):
    c = a + b
    print(c)
sum(5,6,7,8)
```

```python
def sum(a,*b):
    c = a + b
    print(c)
sum(5,6,7,8)
```

```python
def sum(a,*b): # 1st argument is fixed but for 2nd argument
   
    print(type(a))
    print(type(b))

    print(a)
    print(b)
    
sum(5,6,7,8) # Here when * is added for 2nd variable it considers other variable together and displays (6,7,8)
```

```python
def sum(a,*b): # 1st argument is fixed & we fetch each value from the tuple & we can add them.
    c = a

    for i in b:
        c = c + i 
    print(c)
sum (5,6,7,8)
```

```python
def sum(a,*b): # 1st argument is fixed & we fetch each value from the tuple & we can add them.
    c = a

    for i in b:
        c = c + i 
    print(c)
sum (5,6,7,8,9,10, -1, -2, 100)
```

# Kwags (Keyword + Arguments): 

# *args collects extra positional arguments as a tuple.
# **kwargs collects extra keyword arguments as a dictionary.

```python
def person(name, *data):
    print(name)
    print(data)
person('Alex', 36, 'Jhon', 987767) 

```

```python
def person(name, *data):
    print('name')
    print(data)
    
person('Alex', age = 36, home_place = 'southcity', mob = 987767) 
```

```python
def person(name, **data):
    print(name)
    print(data)

person('ALEX', age = 36, home_place ='southcity', mob =987767)
```
