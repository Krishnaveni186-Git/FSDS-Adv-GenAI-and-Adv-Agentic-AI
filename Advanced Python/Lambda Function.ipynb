**Lambda Function**

A lambda function in Python is a small, anonymous (unnamed) function that can have any number of arguments but only **one expression.**

- Avoid using lambda functions for complex logic. In such cases, use a regular def function for better readability.

**Syntax**

lambda arguments: expression




**When to Use Lambda Functions**
- You need a short, simple function.
- The function is used only once.
- It is passed as an argument to functions like:
    - map()
    - filter()
    - sorted()
    - reduce() (from functools)


**Key Point:**
A lambda function is simply a concise way to write a small function: lambda x: x * 2

```python
# Example 1: Simple Lambda Function

square = lambda x: x * x

print(square(5))
```

```python
# Example 2: Add Two Numbers

add = lambda a, b: a + b

print(add(10, 20))
```

```python
# Example 3: Find Maximum of Two Numbers

maximum = lambda a, b: a if a > b else b

print(maximum(15, 8))
```

```python
# Example 4: Using Lambda with map()

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

```python
# Example 5: Using Lambda with filter()

numbers = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, numbers))

print(even)
```

```python
# Example 6: Using Lambda with sorted()

students = [("Alice", 85), ("Bob", 75), ("Charlie", 90)]

sorted_students = sorted(students, key=lambda x: x[1])

print(sorted_students)
```
