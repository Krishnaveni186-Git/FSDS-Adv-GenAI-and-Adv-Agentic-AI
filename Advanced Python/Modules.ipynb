
**Modules**

**What are Modules in Python?**

A **module** in Python is a file containing Python code (functions, variables, classes, or statements) that can be reused in other Python programs. Modules help organize code and avoid rewriting the same code.

**Why use modules?**

- Reuse code
- Organize programs
- Make code easier to maintain
- Share functionality across multiple programs

**Types of Modules**

- **Built-in Modules** – Already provided by Python.
- **User-defined Modules** – Created by the programmer.
- **Third-party Modules** – Installed using pip (e.g., numpy, pandas).

**Common Built-in Modules**




**Advantages of Modules**
- Code reusability
- Better organization
- Easier debugging
- Improved readability
- Faster development

```python
# **1. Built-in Module Example**
     
import math

print(math.sqrt(25))
print(math.factorial(5))
print(math.pi)
```
