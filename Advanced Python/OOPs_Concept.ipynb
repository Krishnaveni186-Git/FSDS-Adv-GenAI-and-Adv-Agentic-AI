Whats is OOPs?
OOPs (Object-Oriented Programming) is a way of writing programs by organizing code around objects.

- 1. Class and Object

Class = Blueprint 

A class is a design or blueprint used to create objects. (Class is collection of objects)

- Object = Real Thing 

An object is created from a class.

- Example of class:
- User Define class
- Sometimes the built-in classes are not enough for our application's requirements.Then we create our own class.This is called a user-defined class.

```python
def greet():
    print('Welcome')
greet()
```

```python
class student:
    def study(self):
        print('Student is studying')

student1 = student()   # Student → User-defined class
                       # study() → Method
student1.study() 
```

```python
class computer:
    def config():
        print('i7', '8gb')

com1 = computer()   # Computer → User-defined class
                    # config() → Method
computer.config()
```

```python
class wish:
    def greet():
        print('Welcome')
w = greet()
#wish.greet()
```

```python
class computer:
    def config():
        print('i7', '8gb')

com1 = computer()   # Computer → User-defined class
                    # config() → Method
computer.config()
com1
```

```python
class computer():
    def config():
        print('i7', '8gb')

com1 = computer()
com1.config()
```

# Using SELF keyword

# Self: self represents the current object.

```python
class computer():
    def config(self):
        print('i7', '8gb')

com1 = computer()
com1.config()
```

```python
class computer():
    def config(self): 
        print('i7', '8gb')

com1 = computer()
com2 = computer()

com1.config()
```

```python
class computer():
    def config(self): 
        print('i7', '8gb')

com1 = computer()
com2 = computer()

com1.config()
com2.config()
```

```python
class computer():     # computer is the class
    def config(self):   # config is the methos   
        print('i5', '8gb')

com1 = computer() 
com2 = computer()
com3 = computer()

com1.config()
com3.config()
com2.config()

print(id(com1))  # Com1 has a stored memmory in 1 object
print(id(com2))  # com2 has a stored memory  in 2 object
print(id(com3))  # com2 has a stored memory  in 3 object ,  therefore all objects store in different memory
```

- **Method:** Methos is also called as special methos (magic/ duder method)
- Methods with double underscores are called magic methods or dunder methods.
- __init__
- __add__
- __len__
- __str__

**Constructer**: A constructor automatically runs when an object is created.

```python
Method Examples
```

```python
class computer():     # computer is the class
    def config(self):   # config is the methos   
        print('i5', '8gb')
com1 = computer()
com1.config()
```

```python
class computer():     # computer is the class
    def __init__(self):   # config is the methos   
        print('i5', '8gb')

com1 = computer()
com2 = computer()
com3 = computer()

print(id(com1))
print(id(com2))
```

```python
class computer():     
    def config(self):   
        print('i5', '8gb')
com1 = computer()

class computer():
    def __init__(self):   
        print('i5', '8gb')
com2 = computer()
```

```python
class computer():

    def config(self):
        print('i5', '8gb')
    
    def __init__(self):
        print('i5', '8gb')

com1 = computer()    
```

```python
class computer():

     def __init__(self):
        print('i7', '16gb')

     def config(self):
        print('i5', '8gb')
    
com1 = computer()
com2 = computer()
com1.config()
```
