Recursion 

Recursiob is a technique where a function calls itself to solve a smaller version of the same problem.

Every recursive function has two important parts:

Base Case – The stopping condition.
Recursive Case – The function calls itself with a smaller problem.

Why do we need Recursion?

Recursion is useful when problems naturally break down into smaller versions of themselves, such as:

- Calculating factorials
- Fibonacci sequence
- Searching folders inside folders
- Tree and graph traversal
- Solving mazes
- Divide-and-conquer algorithms (like Merge Sort and Quick Sort)

Advantages
- Cleaner and shorter code.
- Easier to solve problems that are naturally repetitive.
- Excellent for tree and graph structures.

Disadvantages
- Can use more memory because each function call is stored on the call stack.
- May be slower than loops for simple repetitive tasks.
- Missing the base case leads to a RecursionError.


Recursion vs Loop




Rule of Thumb
Use iteration for simple repetition.
Use recursion when the problem has a recursive structure (a problem made of smaller versions of itself).


The basic syntax of a recursive function in Python is:

def function_name(parameters):
    # Base case (stopping condition)
    if condition:
        return value

    # Recursive case (function calls itself)
    return function_name(smaller_problem)


How the Syntax Works:

def factorial(n):          # 1. Define the function
    if n == 1:             # 2. Base case
        return 1

    return n * factorial(n - 1)   # 3. Recursive call

- def factorial(n): → Defines the function.
- if n == 1: → Checks whether the stopping condition (base case) is reached.
- return 1 → Stops the recursion.
- factorial(n - 1) → Calls the same function with a smaller value.

How Recussion Works:  Example of Factorial Method:

factorial(5)
= 5 × factorial(4)

= 5 × 4 × factorial(3)

= 5 × 4 × 3 × factorial(2)

= 5 × 4 × 3 × 2 × factorial(1)

= 5 × 4 × 3 × 2 × 1

= 120

```python
#  Example 1: Print Numbers from 5 to 1

def print_numbers(n):
    if n == 0:          # Base case
        return

    print(n)
    print_numbers(n - 1)   # Recursive call

print_numbers(5)
```
