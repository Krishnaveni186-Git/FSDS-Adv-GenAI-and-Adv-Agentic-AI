**Natural Language Understanding**

Natural Language Understanding (NLU) is a subfield of Natural Language Processing that provides machines with the ability to interpret and extract meaning from human language. 

**Key Models and Techniques in NLU**

**Transformers**: Modern NLU is powered by transformer architectures that capture contextual relationships - 


- **BERT** (Bidirectional Encoder Representations from Transformers): BERT is trained on large amounts of unlabeled text to learn contextual                                                                                   representations of words based on their surrounding context.Uses bidirectional                                                                           attention to understand sentence meaning.

- **T5** (Text-to-Text Transfer Transformer): Treats every task as text-to-text, which simplifies fine-tuning.

- **GPT**(Generative Pre-trained Transformer): Focuses on generating and understanding text in a conversational setting.

- **RoBERTa** (Robustly Optimized BERT Pretraining Approach): An optimized version of BERT that improves pretraining by removing NSP and using more                                                                     training data.

- **DistilBERT**: DistilBERT is a distilled version of BERT meaning it is trained using **knowledge distillation** a technique where a smaller model                       (student) learns from a larger model (teacher).

  



- **ALBERT** (A Light BERT for Supervised Learning): It will have low paramenters.The goal of this paper to improve the training and results of BERT                                                          architecture by using different techniques like parameter sharing, factorization of embedding                                                            matrix, Inter sentence Coherence loss.

- **XLNet**(Autoregressive Pretraining for Language Understanding): It combines the ideas of BERT and Time Series to autoregressive Model.
