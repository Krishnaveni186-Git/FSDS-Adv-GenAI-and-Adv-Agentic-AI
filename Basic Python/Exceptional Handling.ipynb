**Exception handling**: is a way to handle errors that occur while a Python program is running, so the program does not stop suddenly.
 Python uses try, except, else, and finally for exception handling.

**Main keywords**:
- try – Code that may cause an error
- except – Handles the error
- else – Runs if no error occurs
- finally – Runs whether an error occurs or not

**Simple definition**: Exception handling is the process of handling runtime errors in a program without abruptly stopping its execution.
