```python
plt.figure(figsize = (8, 6))
```

```python
# Scatter plot 

sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'time', size = 'size', palette = 'deep')
plt.title('Scatterplot of Total bill vs Tips')
plt.show()
```

```python
# Line plot

sns.lineplot(data = tips, x = 'size', y = 'total_bill', hue = 'sex', markers = 'o')
plt.title('Lineplot of Total bill vs Size')
plt.show()
```

```python
sns.lineplot(data = tips, x = 'size', y = 'total_bill', hue = 'sex', ci = None, markers = 'o')
plt.title('Lineplot of Total Bill vs Size')
plt.show()
```

```python
tips.columns
```

```python
# bar plot

sns.barplot(data = tips, x = 'day', y = 'total_bill', hue = 'sex', palette = 'muted')
plt.title('Barplot of Total bill by Days')
plt.show()
```

```python
tips.columns
```

```python
# Box plot

sns.boxplot(data = tips, x = 'day', y = 'tip', hue = 'smoker', palette = 'Set2')
plt.title('Boxplot of Tip by Day and Smoker Status')
plt.show()
```

```python
# violin Plot

sns.violinplot(data = tips, x = 'day', y = 'total_bill', hue = 'time', palette = 'pastel', split = True)
plt.title('Violin Plot by Day and Total Bill')
plt.show()
```

```python
tips.columns
```

```python
# Count plot

sns.countplot(data = tips, x = 'day', hue = 'smoker', palette = 'dark')
plt.title('Count Plot by Tips and Day')
plt.show()
```

```python
tips.columns
```

```python
# regression plot

sns.regplot(data = tips, x = 'total_bill', y = 'tip', scatter_kws = {'s': 50}, line_kws = {'color':'red'})
plt.title('Regression Plt of Total Bill vs Tips')
plt.show()


# scatter_kws means "scatter plot keyword arguments".
# s controls the size of the points.


# line_kws means "line keyword arguments".
# Changes the appearance of the regression line.
```

```python
# Histogram of total bill with KDE

sns.histplot(data = tips, x = 'total_bill', bins = 20, kde = True, color = 'blue')
plt.title('Histogram of total bill with KDE')
plt.show()

# KDE stands for Kernel Density Estimation.

# Histogram: Groups data into bins and shows frequencies as bars.
# KDE: Draws a smooth curve showing where data values are concentrated.
```

```python
# pairplot

sns.pairplot(data = tips, hue = 'sex', vars = ['total_bill', 'tip', 'size'], palette = 'husl')
plt.suptitle('pair plot: numeric variables by gender', y = 1.02)
plt.show()

# A pair plot is a collection of scatter plots and distributions that helps you visualize relationships between multiple numerical variables at once.
# Off-diagonal plots → Scatter plots showing relationships between pairs of variables.
# husl provides bright, distinct colors that are easy to differentiate.
# y = 1.02 moves the title slightly upward so it doesn't overlap with the plots.
```

```python
#  catplot 

sns.catplot(data = tips, x = 'day', y = 'tip', hue = 'sex', kind = 'point', palette = 'bright')
plt.title('catplot(point):Tips by day and gender')
plt.show()

# sns.catplot() is used to visualize the relationship between a categorical variable and a numerical variable.
# A point plot shows:
# The average value of tip for each day.
# Points connected by lines.
```

```python
# Jointplot

sns.jointplot(data = tips, x = 'total_bill', y = 'tip', kind = 'scatter', hue = 'smoker', color = 'purple', palette = 'coolwarm')
plt.suptitle('Jointplot: Total Bill vs Tip',  y=1.02)
plt.show()
```

```python
# Facetgrid

g = sns.FacetGrid(tips, col='time', row='smoker', margin_titles=True).map(sns.histplot, 'total_bill', bins=20, kde=True).add_legend()
g
```

```python
#13. strip plot : A Strip Plot displays all individual data points for a categorical variable.

sns.stripplot(data=tips, x='day', y='tip', hue='sex', jitter=True, palette='Set1')
plt.title("strip plot: Tips by data and gender")
plt.show()
```

```python
 # 14. KDE PLOT : Kernal Density Estimation: Reavels distribution speard and shape of the distribution.

sns.kdeplot(data=tips, x='total_bill',hue='sex', fill=True, palette='tab10')
plt.title("kde plot:Total bill density by gender")
plt.show()
```
