**Clustering**

Clustering: Clustering is an unsupervised machine learning technique used to group similar data points into clusters. Unlike classification, clustering does not require labelled data. The algorithm finds hidden patterns or structures in the dataset based on similarity.

**Types of Clustering Algorithms**

1. Partition-Based Clustering

This method divides data into K clusters, where each data point belongs to exactly one cluster.

Popular Algorithm: K-Means
**K-Means Clustering** groups similar data points into clusters without needing labeled data. It is used to uncover hidden patterns when the goal is to organize data based on similarity.

How it Works
- Choose the number of clusters (K).
- Randomly initialize K centroids.
- Assign each data point to the nearest centroid.
- Recalculate the centroid of each cluster.
- Repeat until centroids no longer change significantly.












What is a **Centroid**?

A **centroid** is the **centre point of a cluster** in the K-Means clustering algorithm. It represents the average position of all the data points in that cluster.

The centroid is not always an actual data point. It is calculated by taking the mean (average) of all the points in the cluster.

What is a **New Centroid**?

During K-Means clustering, after assigning data points to the nearest centroid, the algorithm **recalculates the centroid**  based on the newly formed clusters. This updated centre is called the **new centroid**.

The algorithm repeats this process until the centroids stop changing.





2. Hierarchical Clustering

Hierarchical clustering creates a tree-like structure called a dendrogram.


3. Density-Based Clustering

This method groups data based on high-density regions and identifies sparse regions as noise.

Popular Algorithm: DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

4. Model-Based Clustering

Assumes data is generated from a mixture of probability distributions.

Popular Algorithm: Gaussian Mixture Model (GMM)

Instead of assigning each point to exactly one cluster, GMM gives probabilities.


**Elbow Method for optimal value of k in KMeans**

The **Elbow Method** is used to find the optimal number of clusters (k) in K-Means by analyzing how the clustering performance changes with different k values.

Plots WCSS (Within-Cluster Sum of Squares) against different k values
WCSS decreases as k increases, showing better fit
The “elbow point” indicates the optimal k where improvement slows down



**Summary**

- **Clustering** groups similar data without using labels.
- It is an **unsupervised learning** technique.
- **K-Means** is the most popular and fastest method but requires choosing the number of clusters.
- **Hierarchical Clustering** builds a dendrogram and does not require specifying the number of clusters upfront.
- **DBSCAN** is excellent for detecting clusters of arbitrary shapes and identifying outliers.
- **Gaussian Mixture Models** and **Fuzzy C-Means** support overlapping clusters through soft assignments.
- The choice of clustering algorithm depends on the dataset size, expected cluster shapes, presence of noise, and whether the number of clusters is      known.
