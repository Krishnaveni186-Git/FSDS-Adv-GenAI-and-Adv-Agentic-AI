# Grid Search CV and Random Search CV

**Grid Search CV and Random Search CV** are **hyperparameter tuning** techniques used to find the best combination of model parameters that gives the highest performance.

**Lets Understand Parameter and Hyperparameter**

**Parameters**: Learned automatically by the model during training (e.g., weights in linear regression).

**Hyperparameters**: Set by the user before training (e.g., number of trees, learning rate, maximum tree depth).


What is Cross-Validation (CV)?

**Cross-Validation** is a technique used to evaluate a model more reliably.

The most common method is **K-Fold Cross-Validation**:

- Split the dataset into K equal parts (folds).
- Train on K−1 folds.
- Test on the remaining fold.
- Repeat until every fold has been used as the test set.
- Average all the scores.

Example for 5-Fold CV

Fold 1 → Test
Fold 2-5 → Train

Fold 2 → Test
Fold 1,3,4,5 → Train

...
Average Accuracy = Final Score

This gives a more reliable estimate than using a single train-test split.




**Grid Search CV**: Grid Search CV tries **every possible combination** of hyperparameters.

Suppose you are tuning a Random Forest model.

Hyperparameters:

n_estimators = [100, 200]

max_depth = [5, 10]

min_samples_split = [2, 5]

**Total combinations**:
2 × 2 × 2 = 8

If using 5-Fold Cross Validation:
8 combinations × 5 folds = 40 model trainings

Grid Search trains **all 40 models** and chooses the best one.

min_samples_split → [2,5]

**Random Search CV**: Random Search CV does not test every combination.

Instead, it randomly selects a specified number of combinations from the search space.

**Example**

Suppose the search space is:

n_estimators = [100,200,300,400,500]

max_depth = [5,10,15,20]

learning_rate = [0.01,0.05,0.1,0.2]

Total combinations:

5 × 4 × 4 = 80

Instead of trying all 80 combinations, Random Search might evaluate only **10 random combinations**.

**How Random Search Works**

All Possible Combinations

       │
       ▼

Randomly Select N Combinations

       │
       ▼

Perform K-Fold CV

       │
       ▼

Average Score

       │
       ▼

Best Random Combination





**Summary**

**Grid Search CV** systematically evaluates every hyperparameter combination using cross-validation, making it thorough but computationally expensive.

**Random Search CV** evaluates only a random subset of combinations, making it much faster while often finding near-optimal settings.

For **small search spaces**, Grid Search is a good choice. For **large or complex search spaces**, Random Search is usually more efficient.
