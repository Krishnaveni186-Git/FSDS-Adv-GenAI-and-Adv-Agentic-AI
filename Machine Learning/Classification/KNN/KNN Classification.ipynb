**K- Nearest Neighbour**

K-Nearest Neighbors (KNN) Classification is a **supervised machine learning algorithm** used to classify data points based on the classes of their nearest neighbors.




**How KNN Works**
- Choose the number of neighbors (K).
- Calculate the distance between the new data point and all training data points.
- Common distance metric: Euclidean distance
- 




- Select the K nearest neighbors.
- Count the class labels of these neighbors.
- Assign the new data point to the class with the majority vote.


**Advantages**
- Simple and easy to understand.
- No training phase (lazy learning).
- Works well for small datasets.
- Can classify multi-class problems.

**Disadvantages**
- Slow for large datasets because distances are computed for all training points.
- Sensitive to irrelevant features and outliers.
- Performance depends on the choice of K and the distance metric.
- Requires feature scaling (e.g., normalization or standardization).

**Applications**
- Image recognition
- Handwriting recognition
- Medical diagnosis
- Recommendation systems
- Pattern recognition
- Fraud detection

**Summary**
**Type**:     Supervised learning algorithm
**Purpose**:  Classification (and can also be used for regression)
**Principle**: Classifies a new data point based on the majority class among its **K nearest neighbors**
**Common Distance Metric**: Euclidean distance

KNN is one of the simplest and most intuitive classification algorithms, making it a popular choice for introductory machine learning and problems with relatively small datasets.
