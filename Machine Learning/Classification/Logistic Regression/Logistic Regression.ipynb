**Logistic regression**

**Logistic regression** is a supervised machine learning algorithm used for **classification** problems, where the target variable is categorical (e.g., Yes/No, 0/1, Spam/Not Spam).








The predicted probability is then compared to a threshold (commonly **0.5**) to assign a class:

If probability ≥ 0.5 → Class 1
If probability < 0.5 → Class 0

**Steps in logistic regression**
- Collect and preprocess data.
- Fit the logistic regression model to the training data.
- Estimate the probability of each class.
- Apply a decision threshold to classify observations.
- Evaluate the model using metrics such as accuracy, precision, recall, F1-score, ROC-AUC, and a confusion matrix.

**Assumptions**
- The dependent variable is binary (for binary logistic regression).
- Observations are independent.
- There is a linear relationship between the predictors and the log-odds of the outcome.
- Little or no multicollinearity among predictors.
- Large sample size is preferred.

**Types of logistic regression**
- Binary logistic regression: Two classes (e.g., Pass/Fail).
- Multinomial logistic regression: More than two unordered classes (e.g., Red, Green, Blue).
- Ordinal logistic regression: Ordered classes (e.g., Low, Medium, High).

**Applications**
- Email spam detection
- Disease diagnosis
- Credit risk prediction
- Customer churn prediction
- Fraud detection
- Sentiment analysis

**ROC-AUC**
**ROC (Receiver Operating Characteristic) Curve**

The ROC curve plots:

**True Positive Rate (Recall)** on the Y-axis
**False Positive Rate (FPR)** on the X-axis

**FPR = FP / FP + TN**

The curve shows how the classifier performs at different probability thresholds.

**AUC (Area Under the Curve)**

AUC measures the **overall ability** of the model to distinguish between positive and negative classes.

Values range from **0 to 1**.





**Example**

Suppose:

Model A: AUC = 0.95
Model B: AUC = 0.72

Model A is much better at distinguishing positive from negative cases.
