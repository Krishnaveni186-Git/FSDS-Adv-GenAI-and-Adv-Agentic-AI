** Naive Bayes Algorithm in Classification**

**Naive Bayes** is a supervised machine learning algorithm used mainly for **classification problems**.

It is based on **Bayes' Theorem** and assumes that all input features are **independent of each other**. This assumption is called “naive”, which is why the algorithm is named Naive Bayes.

 **Bayes Theorem** 



**Key Concepts of Naive Bayes**

**Conditional Probability**:




**Independence Assumption**: This is the most important concept. Naive Bayes assumes that all features are independent of each other.

**Prior Probability**: This is the probability of a class before looking at the features.

**Likelihood**: This is the probability of observing a feature given a particular class.

**Posterior Probability**: This is the final probability calculated after considering the features.


**Types of Naive Bayes**

**Gaussian Naive Bayes**: Used when numerical features follow approximately a normal distribution.

**Multinomial Naive Bayes**: Commonly used for text classification.

**Bernoulli Naive Bayes**: Used when features are binary YES or NO Present or Absent

**Advantages**
- Simple and easy to implement
-  Very fast for training and prediction
-  Works well with high-dimensional data
-  Excellent for text classification
-  Requires relatively little training data

**Disadvantages**
- Assumes features are independent
- This assumption may not be true in real-world data
- Can suffer from the **zero-frequency problem**
