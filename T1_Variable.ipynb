# 1. Sorting and Printing variable

```python
X=10    # Variable X stores the values as 10

print(X) # Priting the variable X displays the value 10
```

# 2. Using variables in Expressions

```python
# Assigning varibles to value

a=5
b=3
# adding 2 variables and sorting result as 'result'
result= a+b
# printing the result

print (result)
```

# 3 Changing the value of variable

```python
score = 50  # Initial value
print (score) # output is 50

# Changing the value of score
score = 100
print (score) # output is 100
```

# 4. Concatenating Strings

```python
# Assigning value to variable

first_name = 'John'
last_name = 'Doe'

# Concatenating the variable first_name and last_name as 'full_name'

full_name = first_name + " " + last_name

print(full_name)
```

# 5. Using Variable in Calculation

```python
# Assigning values to variables
length = 10
width = 5   

# Calculating the area of rectangle

area = length * width
print(area)
```

# Reassigning values to variables

# Python allows you to assign new value to variable anytime

```python
x = 10 # initial value assigned to variable
print(x) # output is 10

# Reassigning value to variable as 20

x = 20
print(x) # New value is 20
```
